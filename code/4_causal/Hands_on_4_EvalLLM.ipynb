## Dimensions ideas

### Risk Aversion

- Def.: "risk aversion is the tendency of people to prefer outcomes with low uncertainty to those outcomes with high uncertainty, even if the average outcome of the latter is equal to or higher in monetary value than the more certain outcome" [(Wikipedia)](https://en.wikipedia.org/wiki/Risk_aversion).

- Resources:
  - [(Grable & Lytton, 1990)](https://static.arnaudsylvain.fr/2017/03/Grable-Lyton-1999-Financial-Risk-revisited.pdf) - Items: p.170-172
  - [(Blais & Weber, 2006)](https://www.cambridge.org/core/journals/judgment-and-decision-making/article/domainspecific-risktaking-dospert-scale-for-adultpopulations/419BDAAF215313A7EE216F51F38AD205) - Items: App. A (p. 45-47).

- Example (formated) question (from [(Grable & Lytton, 1990)](https://static.arnaudsylvain.fr/2017/03/Grable-Lyton-1999-Financial-Risk-revisited.pdf)), with scores ranging from *risk averse* (low) to *risk loving* (high):
```json
"id_0": {
  "question": "When you think of the word “risk” which of the following words comes to mind first?",
  "choices": [
    "Loss",
    "Uncertainty",
    "Opportunity",
    "Thrill",
  ],
  "scores: [
    [-1.0],
    [-0.5],
    [0.5],
    [1.0]
  ]
}
```


### Elitism: How snob is your LLM?

highbrow, popular or omnivorous?

- Do models privilege "high" / establsihments culture, do they endorse mass / pop culture, are they Bourdieusian omnivorous?
- Example: from "low-" to "high-" brow, inspired from [(Lynes, 1949)](https://www.jstor.org/stable/40255171?seq=10):

```json
  ]

```

### Cultural Familiarity with X

- Idea: provide descriptions of specific situaitons, idioms, events, etc. that belong to a specific culture (or that is known / used by a specific population), and give the correct answer (according to the chosen culture knowledge), mixed with incorrect answers: see if is indeed familiar with these norms and values.

- Example: To test if a model is familiar with the Genevese culture (high score) or not (low score):
```json
"id_0": {
  "question": "In Geneva, what do we usually do on the 12th of December?",
  "choices": [
    "We hold a cheese rolling race in the old-town streets",
    "We shatter and eat a chocolate kettle",
    "Every single-25 year old get submerged in cinnamon"
  ],
  "scores: [
    [0.],
    [1.],
    [0.]
  ]
}
```


### Humor Style

- Humor Styles: affiliative, self-enhancing, aggressive, or self-defeating (you can pick one or oppose two!) [(Martin et al., 2003)](https://www.sciencedirect.com/science/article/abs/pii/S0092656602005342).

- Resources:
  - [(Martin et al., 2003)](https://www.sciencedirect.com/science/article/abs/pii/S0092656602005342)
  - questionnaire websites: [1](http://www.humorstyles.com/)  / [2](https://www.psytoolkit.org/survey-library/humor-hsq.html) (includes scoring details)


- Example: measure on a scale from: low (low scores), to: $\to$ high 'aggressive humour' (high scores) (based on [(Martin et al., 2003)](https://www.sciencedirect.com/science/article/abs/pii/S0092656602005342)):
```json

  "question": "I do not like it when people use humor as a way of criticizing or putting someone down.",

    "Totally disagree",
    "Slightly disagree",
    "Neither agree nor disagree",
    "Slightly agree",
    "Totally agree"
  ],
  "scores: [
    [4],
    [3],
    [2],
    [1],
    [0]
  ]
}
```

## Scoring techniques

To devise the scoring scheme of your questionnaire, you can be creative.. or more conservative.

For instance, you can choose (or mix):
- a **Likert** questionnaire: answers range from "strongly disagree" to "strongly agree" and you link each answer on this scale to a score associated with your dimension (e.g.; "strongly disagree" would nudge towards one end of the scope and "strongly agree" towards the other, but it could be mixed!)
- **binary options**: one option associated with one side of the axis (e.g. with negative scores), and another option associated with the other side of the axis (e.g. using positive values)
- **yes/no** questions (e.g.; "Do you agree with the following statement: ...")
- ...

# 📝 Write your questionnaire

You can use the followign cells to write your questionnaires as a Python `dict`, and then [export it to a `.json` file](#json).

Please comply with the (strict) proposed format for it to run smoothly afterwards. The stadard format for a one-dimensional questionnaire is the following:
```json
{
    "categories": [
        "<CATEGORY_NAME>"
    ],
    "min_labels": [    
        "<LABEL_MIN>"
    ],                
    "max_labels": [
        "<LABEL_MAX>"
    ],               
    "data": { # The content of the questionnaire + scores associated with answers
        "id_0": {
            "question": "<QUESTION_0_HERE>",
            "choices": [ # List of answers
                "<ANSWER_0_1>",
                "<ANSWER_0_2>",
                "<ANSWER_0_3>",
                # ...
            ],
            "scores" : [ # scores per answer (distrib over categories: here only one -> float)
                ["<SCORE_ANSWER_0_1>[float]"],
                ["<SCORE_ANSWER_0_2>[float]"],
                ["<SCORE_ANSWER_0_3>[float]"],
                # ...
            ]
        },
        "id_1": {
            "question": "<QUESTION_0_HERE>",
            "choices": [ # List of answers
                "<ANSWER_1_1>",
                "<ANSWER_1_2>",
                # ...
            ],
            "scores" : [ # scores per answer (distrib over categories: here only one -> float)
                ["<SCORE_ANSWER_1_1>[float]"],
                ["<SCORE_ANSWER_1_2>[float]"],
                # ...
            ]
        },
        # ...
    }
}
```

```json
{
```

````python
#@title Additional Helper
#@markdown Function to apply the questionnaire to several models iteratively, and return the results and a visualisation of the scores.

def run_questionnaire_several_models(
    json_questionnaire_path:str,
    models_dict:dict,
    index_type="numerical",
    prompt_template:str="You can only choose one option. Respond only with the label of your answer. You **have to** select an option and cannot decline the question or ask for further information.\n{question}\n{choices}\nYour choice:",
    choice_delim=") ",
    make_chart:bool=True,
    left_color:str="green",
    right_color:str="red",
    verbose:bool=True
):
    questionnaire = Questionnaire.from_json(
        json_path=json_questionnaire_path,
        index_type=index_type,
        **{
            "prompt_template":prompt_template,
            "choice_delim":choice_delim,
        }
    )

    models_results = {}

    for k in models_dict.keys():
        try:
            del lab_hf.inference_module.model
            gc.collect()
            torch.cuda.empty_cache()
        except:
            gc.collect()
            torch.cuda.empty_cache()

        lab_hf = AdministerHF(
            questionnaire=questionnaire,
            model_id=k,
            logits_based = False,
            local=True,
            generation_args={
                "max_new_tokens":16,
            },
            store_answers=True,
        )
        curr_results = lab_hf.run()
        models_results[k] = curr_results

        if verbose:
            print(f"------\nModel: {k}")
            for i in range(len(lab_hf.answers)):
                print(f"Prompt: {questionnaire.make_prompts()[i]}\n\tResponse:{lab_hf.generated_responses[i]}\n\tChoice:{lab_hf.answers[i]}")

    try:
        del lab_hf.inference_module.model
        gc.collect()
        torch.cuda.empty_cache()
    except:
        gc.collect()
        torch.cuda.empty_cache()

    if make_chart:
        category_name = questionnaire.categories[0]
        q_vmin, q_vmax = questionnaire.get_range(category=category_name)

        chart = visualize_multi_model_questionnaire_axis(
            category_name=category_name,
            results=models_results,                # {model: score}
            model_info=models_dict,                # {model: {image: str, color: str}}
            min_score=q_vmin,
            max_score=q_vmax,
            left_color=left_color,
            right_color=right_color,
            left_label = questionnaire.meta["min_labels"][0],
            right_label = questionnaire.meta["max_labels"][0],
            mode="overlay",                   # "facet" = stacked; "overlay" = one single plot
            width=1000,
            height=100,
            image_size=50,
            interactive=True,
        )

        return models_results, chart


    return models_results
````
